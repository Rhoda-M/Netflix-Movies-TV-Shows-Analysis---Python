```python
# Spacing - Type Column
df["Type"] = df["Type"].str.strip()
```

```python
df[df["Rating"].str.contains("min", case=False, na=False)]
```

```python
df.loc[
    df["Rating"].str.contains("min", case=False, na=False),
    ["Title", "Rating", "Duration"]
]
```

```python
# Fixing incorrect records - Moving Duration values from Rating to Duration
mask = df["Rating"].str.contains("min", case=False, na=False)

df.loc[mask, "Duration"] = df.loc[mask, "Rating"]
df.loc[mask, "Rating"] = "Not Specified"
```

```python
df[df["Rating"].str.contains("min", case=False, na=False)]
```

```python
df["Rating"].value_counts()
```

```python
# Removing unnecessary spaces
df["Rating"] = df["Rating"].str.strip()
```

```python
# Standardizing capitalization
df["Rating"] = df["Rating"].str.upper()
```

```python
df
```

```python
# Cleaning other text columns 
text_columns = ["Title", "Director", "Cast", "Country", "Rating", "Duration", "Genre", "Type"]

for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
# Exporting cleaned dataset
df.to_csv("netflix_cleaned.csv", index=False)
```
